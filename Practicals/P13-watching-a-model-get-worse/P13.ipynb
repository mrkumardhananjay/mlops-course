2. Detect when incoming data stops resembling the training data.
3. Write an alert rule that does not fire on ordinary noise.
4. Audit the model by group and establish who it serves worst.

## What this practical is about

Every practical so far has produced a model and stopped. This one begins after
deployment, which is where the error that matters appears.

The failure examined here is **silent**. No exception is raised, no status code
changes, and the service remains fast and available. Only the answers become
wrong, and they become wrong for a particular group of customers rather than
for everybody.

Two signals are available, and they differ in an important respect.

- **Input drift** is computable the moment a request arrives, because it needs
  only the inputs. It is the early warning.
- **Error** requires knowing what actually happened, which for a delivery
  arrives roughly an hour later. It is the verdict.

The practical also separates two things that are frequently conflated.
**Data drift** means the input distribution moved. **Concept drift** means the
relationship between inputs and the true answer moved. The simulation below
produces both, deliberately, and Step 5 distinguishes them from the data
alone.

## Step 0b: ensure the shared dataset exists

Every practical in this course uses the same 600 synthetic delivery
observations. The next cell confirms the file is present and reconstructs it
from the course seed if it is absent.

```python
# The delivery-time dataset every practical in this course uses.
# If the file is missing we build it again from the same seed, so every
# student in the room gets byte-for-byte the same 600 rows.

import csv
from pathlib import Path

import numpy as np

SEED = 42
N_ROWS = 600
from pathlib import Path

def find_data(name="delivery_times.csv"):
    """Locate the shared dataset from any working directory.

    Walks upward from the current folder looking for data/<name>, so the
    notebook works whether it is launched from the practical folder, the
    Practicals folder, or the repository root."""
    here = Path.cwd().resolve()
    for folder in [here, *here.parents]:
        for candidate in (folder / "data" / name,
                          folder / "Practicals" / "data" / name):
            if candidate.exists():
                return candidate.resolve()
    raise FileNotFoundError(
        f"{name} not found in any data/ folder at or above {here}")

DATA = find_data()
print("dataset:", DATA)


def make_delivery_csv(path=DATA):
    """Write the 600-row delivery dataset. Same formula as the lectures."""
    rng = np.random.default_rng(SEED)
    distance_km = np.round(rng.uniform(0.5, 12.0, N_ROWS), 2)
    prep_time_min = np.round(rng.uniform(5, 30, N_ROWS), 0)
    traffic_level = rng.integers(1, 4, N_ROWS)
    rain = rng.binomial(1, 0.25, N_ROWS)
    delivery_min = np.round(
        6.0
        + 3.1 * distance_km
        + 0.65 * prep_time_min
        + 4.2 * traffic_level
        + 5.5 * rain
        + rng.normal(0, 2.5, N_ROWS),
        1,
    )
    path.parent.mkdir(parents=True, exist_ok=True)
    with path.open("w", newline="", encoding="utf-8") as fh:
        w = csv.writer(fh)
        w.writerow(["distance_km", "prep_time_min", "traffic_level",
                    "rain", "delivery_min"])
        for i in range(N_ROWS):
            w.writerow([distance_km[i], int(prep_time_min[i]),
                        int(traffic_level[i]), int(rain[i]), delivery_min[i]])
    return path


if not DATA.exists():
    make_delivery_csv()
    print("dataset rebuilt ->", DATA)
else:
    print("dataset found   ->", DATA)
```

## Step 1: train the model that will go into production

This is the model from P02, trained on data up to the present. In practice it
would be deployed and left. Here it is deployed and then watched.

```python
from pathlib import Path

import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import train_test_split

WORK = Path("work")
WORK.mkdir(exist_ok=True)
FEATURES = ["distance_km", "prep_time_min", "traffic_level", "rain"]

orders = pd.read_csv(DATA)
X_train, X_test, y_train, y_test = train_test_split(
    orders[FEATURES], orders["delivery_min"],
    test_size=0.2, random_state=42,
)
model = LinearRegression().fit(X_train, y_train)

launch_mae = mean_absolute_error(y_test, model.predict(X_test))
print(f"MAE at launch: {launch_mae:.2f} minutes")
print("Deployed. Nobody is watching.")
```

## Step 2: eight weeks of traffic arrive

The cell below simulates what happens next. Weeks 1 to 4 resemble the training
data. From week 5 the company expands into a suburb and trips become longer.

Two distinct things change, and the distinction between them is the theory made
concrete.

**The inputs move.** Distances extend beyond anything present in the training
data. That is **data drift**.

**The relationship also changes.** Beyond roughly 12 km a rider leaves the city
roads, so each additional kilometre out there costs more minutes than a
kilometre in town. The training data contained no such trips, so the model
never learned this. That is **concept drift**.

Read `long_trip_penalty` in the code below. It is zero for every trip under
12 km, which is why weeks 1 to 4 appear entirely normal and why nothing in the
service reports a fault.

```python
SEED = 7
rng = np.random.default_rng(SEED)
ORDERS_PER_WEEK = 150


def week_of_orders(week):
    """One week of live traffic. From week 5, trips get longer."""
    stretch = 1.0 if week <= 4 else 1.0 + 0.22 * (week - 4)

    distance_km = np.round(
        rng.uniform(0.5, 12.0, ORDERS_PER_WEEK) * stretch, 2)
    prep_time_min = np.round(
        rng.uniform(5, 30, ORDERS_PER_WEEK), 0).astype(int)
    traffic_level = rng.integers(1, 4, ORDERS_PER_WEEK)
    rain = rng.binomial(1, 0.25, ORDERS_PER_WEEK)

    # Beyond 12 km the rider is on highways, not city roads, and
    # every extra kilometre costs more. No trip in the training
    # data was that long, so the model has never seen this rule.
    long_trip_penalty = 1.8 * np.maximum(distance_km - 12.0, 0.0)

    true_minutes = np.round(
        6.0 + 3.1 * distance_km + 0.65 * prep_time_min
        + 4.2 * traffic_level + 5.5 * rain
        + long_trip_penalty
        + rng.normal(0, 2.5, ORDERS_PER_WEEK), 1)

    return pd.DataFrame({
        "week": week,
        "distance_km": distance_km,
        "prep_time_min": prep_time_min,
        "traffic_level": traffic_level,
        "rain": rain,
        "actual_min": true_minutes,
    })


live = pd.concat([week_of_orders(w) for w in range(1, 9)],
                 ignore_index=True)
live["predicted_min"] = model.predict(live[FEATURES])

print("orders received:", len(live))
print(live.groupby("week")["distance_km"].mean().round(2)
      .to_string())
```

Examine the mean distance per week. Weeks 1 to 4 sit near 6 km, and from week 5
the figure climbs. That is data drift, and it occurred without anybody
deploying anything.

## Step 3: measure the error week by week

A single MAE for the whole period would conceal the trend entirely. The
measurement is therefore taken per week, as a dashboard would take it.

```python
weekly = (live.assign(
              error=lambda d: (d["actual_min"] - d["predicted_min"]).abs())
          .groupby("week")
          .agg(orders=("error", "size"),
               mae=("error", "mean"),
               mean_distance=("distance_km", "mean"))
          .round(2))

print(weekly.to_string())
print()
print(f"MAE at launch was {launch_mae:.2f} minutes.")
```

Weeks 1 to 4 sit near the launch figure. The error then climbs and continues to
climb. No code changed and no notification was issued, which is the definition
of silent failure.

```python
%matplotlib inline
import matplotlib.pyplot as plt

fig, (left, right) = plt.subplots(1, 2, figsize=(11, 3.6))

left.plot(weekly.index, weekly["mae"], marker="o", color="#9E3B33")
left.axhline(launch_mae, linestyle="--", color="#6B7785",
             label=f"launch MAE {launch_mae:.2f}")
left.axhline(launch_mae * 1.5, linestyle=":", color="#B4703A",
             label="alert threshold")
left.set_title("Prediction error by week")
left.set_xlabel("week")
left.set_ylabel("MAE (minutes)")
left.legend(fontsize=8)

right.plot(weekly.index, weekly["mean_distance"], marker="s",
           color="#2C5F79")
right.set_title("Average distance by week")
right.set_xlabel("week")
right.set_ylabel("km")

plt.tight_layout()
plt.show()
```

## Step 5: establish the direction of the error, and where it begins

MAE discards the sign. A model that is five minutes early and one that is five
minutes late produce the same MAE, and a customer complains about only one of
them.

The sign is therefore retained. **Signed error** is the actual value minus the
predicted value.

- Above 0, the food arrived later than the application promised.
- Below 0, it arrived earlier.

The left panel places every order from all eight weeks against its distance. The
dotted line marks the longest trip present in the training data.

Before running the cell, make a prediction. If the model were simply somewhat
worse everywhere, what shape would the points form?

```python
live["signed_error"] = live["actual_min"] - live["predicted_min"]
longest_trained = X_train["distance_km"].max()
seen = live["distance_km"] <= longest_trained

inside, beyond = live[seen], live[~seen]
inside_mae = inside["signed_error"].abs().mean()
beyond_mae = beyond["signed_error"].abs().mean()
beyond_bias = beyond["signed_error"].mean()

print(f"Longest trip in the training data: {longest_trained:.2f} km")
print()
print(f"{'':<24}{'orders':>7}{'MAE':>8}{'avg signed error':>19}")
print(f"{'inside training range':<24}{len(inside):>7}"
      f"{inside_mae:>8.2f}{inside['signed_error'].mean():>+19.2f}")
print(f"{'beyond it':<24}{len(beyond):>7}"
      f"{beyond_mae:>8.2f}{beyond_bias:>+19.2f}")

weekly_bias = live.groupby("week")["signed_error"].mean()

fig, (left, right) = plt.subplots(
    1, 2, figsize=(11, 3.8), gridspec_kw={"width_ratios": [1.6, 1]})

left.scatter(inside["distance_km"], inside["signed_error"], s=8,
             color="#B9BFC7", label="inside training range")
left.scatter(beyond["distance_km"], beyond["signed_error"], s=10,
             color="#9E3B33", label="beyond it")
left.axhline(0, color="#6B7785", linewidth=1)
left.axvline(longest_trained, color="#6B7785", linestyle=":",
             linewidth=1)
left.text(0.02, 0.95, "above 0: later than promised",
          transform=left.transAxes, fontsize=8, va="top",
          color="#6B7785")
left.set_title("Every dot is one order, weeks 1-8", fontsize=10)
left.set_xlabel("distance_km")
left.set_ylabel("actual - predicted (minutes)")
left.legend(fontsize=8, loc="upper center")
left.spines[["top", "right"]].set_visible(False)

right.bar(weekly_bias.index, weekly_bias.values,
          color=["#9E3B33" if b > 1 else "#6B7785"
                 for b in weekly_bias])
right.axhline(0, color="#6B7785", linewidth=1)
right.set_title("Average signed error by week", fontsize=10)
right.set_xlabel("week")
right.set_ylabel("minutes late (+) or early (-)")
right.spines[["top", "right"]].set_visible(False)

plt.tight_layout()
plt.show()
```

**Reading the left panel**, from left to right.

- Up to the dotted line the grey cloud sits evenly about zero. Within the range
  it was trained on, the model remains as accurate as it was at launch. Compare
  the MAE in the table against the launch figure.
- Beyond the dotted line the red points leave the cloud and climb in a straight
  ramp. Each additional kilometre out there contributes minutes the model has no
  information about. That ramp is the `long_trip_penalty` from Step 2, recovered
  from the data alone.
- Almost every red point lies above zero. The model is not randomly wrong in
  that region. It is wrong in one direction, promising the food earlier than it
  arrives, consistently.

The right panel states the same finding week by week: bias near zero for weeks
1 to 4, after which customers begin waiting longer than promised.

The prediction made before the cell ran can now be checked. A model that is
uniformly somewhat worse produces a wider cloud still centred on zero. What
appears instead is a bend at the boundary of the training range. A model is
trustworthy only within the range it was trained on, which is the finding L29
examines at length.

## Step 6: detect the drift in the inputs

The error chart requires the true delivery times, which arrive an hour late.
The input chart does not. It can be computed the moment a request lands, which
is the entire reason input drift is worth measuring separately.

**PSI**, the population stability index, compares two distributions bucket by
bucket. The training data is divided into ten buckets and the share of new data
falling into each is compared against it.

```python
def psi(reference, current, buckets=10):
    """Population Stability Index between two samples."""
    edges = np.percentile(reference, np.linspace(0, 100, buckets + 1))
    edges[0], edges[-1] = -np.inf, np.inf

    ref_share = np.histogram(reference, bins=edges)[0] / len(reference)
    cur_share = np.histogram(current, bins=edges)[0] / len(current)

    # a tiny floor keeps empty buckets from producing infinity
    ref_share = np.clip(ref_share, 1e-6, None)
    cur_share = np.clip(cur_share, 1e-6, None)

    return float(np.sum((cur_share - ref_share)
                        * np.log(cur_share / ref_share)))


print(f"{'week':<6}{'PSI(distance)':<16}{'reading'}")
print("-" * 48)
for week in range(1, 9):
    current = live.loc[live["week"] == week, "distance_km"]
    score = psi(X_train["distance_km"], current)
    if score < 0.1:
        reading = "stable"
    elif score < 0.25:
        reading = "worth watching"
    else:
        reading = "SHIFTED"
    print(f"{week:<6}{score:<16.3f}{reading}")
```

Two observations here.

PSI crosses 0.25 at week 6, while the error was only beginning to move.
Watching the inputs is the early warning, because it does not require waiting
for the truth to arrive.

Now examine week 3: 0.127, inside the band worth watching, with nothing
actually wrong. That is a false alarm produced by ordinary randomness, and it is
the reason Step 8 does not alert on a single reading.

## Step 8: write an alert rule that does not fire on noise

A threshold alone fires on every noisy week and is muted within a fortnight. A
usable rule requires the condition to **persist**.

The rule used here is that the error must exceed 1.5 times the launch figure for
two consecutive weeks.

```python
def check_alerts(weekly, launch_mae, multiplier=1.5, in_a_row=2):
    """Fire only when the error stays high for several weeks."""
    threshold = launch_mae * multiplier
    streak, alerts = 0, []

    for week, row in weekly.iterrows():
        if row["mae"] > threshold:
            streak += 1
            if streak >= in_a_row:
                alerts.append({
                    "week": int(week),
                    "mae": float(row["mae"]),
                    "threshold": round(threshold, 2),
                })
        else:
            streak = 0
    return alerts


alerts = check_alerts(weekly, launch_mae)

print(f"threshold: {launch_mae * 1.5:.2f} minutes, "
      f"sustained for 2 weeks")
print()
for alert in alerts:
    print(f"  ALERT week {alert['week']}: MAE {alert['mae']:.2f} "
          f"is above {alert['threshold']}")
if not alerts:
    print("  no alerts")
print()
print("First alert:", alerts[0]["week"] if alerts else "none")
```

## Step 9: draw the warning and the alarm on one timeline

Two signals are now available and they arrive at different speeds.

- **PSI**, the upper panel, is available the moment an order arrives. It is the
  warning.
- **MAE**, the lower panel, waits for the true delivery time. Its alert rule is
  the alarm.

In the lower panel a bar is grey when the error is normal, orange when it is
above the threshold but the rule is still waiting to see whether the condition
persists, and red when the alert fires.

```python
from matplotlib.patches import Patch

week_psi = pd.Series({
    w: psi(X_train["distance_km"],
           live.loc[live["week"] == w, "distance_km"])
    for w in range(1, 9)})
threshold = launch_mae * 1.5
alert_weeks = {a["week"] for a in alerts}

first_shift = int(week_psi[week_psi >= 0.25].index[0])
first_alert = alerts[0]["week"]

fig, (top, bottom) = plt.subplots(2, 1, figsize=(9, 5.6),
                                  sharex=True)

top.axhspan(0.10, 0.25, color="#B4703A", alpha=0.15)
top.axhspan(0.25, 1.25, color="#9E3B33", alpha=0.10)
for y, word in [(0.04, "stable"), (0.17, "watching"),
                (0.70, "shifted")]:
    top.text(8.45, y, word, fontsize=8, color="#6B7785",
             va="center")
top.plot(week_psi.index, week_psi.values, marker="o",
         color="#2C5F79")
top.set_ylim(0, 1.25)
top.set_ylabel("PSI(distance)")
top.set_title("Inputs: known the moment an order arrives",
              fontsize=10)

colours = []
for week, mae in weekly["mae"].items():
    if week in alert_weeks:
        colours.append("#9E3B33")
    elif mae > threshold:
        colours.append("#B4703A")
    else:
        colours.append("#B9BFC7")
bottom.bar(weekly.index, weekly["mae"], color=colours)
bottom.axhline(threshold, color="#B4703A", linestyle=":")
bottom.set_ylabel("MAE (minutes)")
bottom.set_xlabel("week")
bottom.set_title("Errors: known only once the true times arrive",
                 fontsize=10)
bottom.legend(handles=[
    Patch(color="#B9BFC7", label="normal"),
    Patch(color="#B4703A", label="above the line, waiting"),
    Patch(color="#9E3B33", label="ALERT fired"),
], fontsize=8, loc="upper left")

for ax in (top, bottom):
    ax.axvline(first_shift, color="#2C5F79", linestyle="--",
               linewidth=1, label=f"warning: week {first_shift}")
    ax.axvline(first_alert, color="#9E3B33", linestyle="--",
               linewidth=1, label=f"alarm: week {first_alert}")
    ax.spines[["top", "right"]].set_visible(False)
top.legend(fontsize=8, loc="upper left")
bottom.set_xticks(range(1, 9))
plt.tight_layout()
plt.show()

print(f"Inputs first SHIFTED:    week {first_shift}")
print(f"Error alert first fired: week {first_alert}")
print(f"Warning came {first_alert - first_shift} week(s) "
      f"before the alarm.")
```

Read the timeline one week at a time.

**Week 3.** The PSI point sits in the orange band while the error bar beneath it
is grey. The inputs moved; the service did not. Alerting on PSI alone would have
woken somebody for nothing.

**Week 6.** PSI enters the red band. The error bar is above the threshold but
orange, because the rule requires a second week before it acts.

**Week 7.** A second consecutive week above the threshold. The bar turns red and
the alarm fires.

That is the two level design teams actually operate. A PSI warning opens a
ticket for somebody to examine during the day, and only the error alarm wakes
somebody at night.

## Step 10: triage, and what to do this week

Each week a monitoring team answers two questions.

1. Did the inputs move? PSI at or above 0.25.
2. Did the error rise? MAE above 1.5 times the launch figure.

Two questions yield four states, and each implies a different action.

| | error normal | error high |
|---|---|---|
| **inputs stable** | healthy, take no action | error without input drift, investigate |
| **inputs moved** | early warning, watch closely | drift confirmed, retrain |

That table is L27's decision table in the form a monitoring rota uses it. The
code places each of the eight weeks into one of the four states, as a table and
as a map.

```python
def triage(psi_score, mae_ratio):
    """One week's monitoring verdict, from two numbers."""
    moved = psi_score >= 0.25
    worse = mae_ratio > 1.5
    if moved and worse:
        return "drift confirmed: retrain"
    if moved:
        return "early warning: watch"
    if worse:
        return "error with no input drift: investigate"
    return "healthy"


report = pd.DataFrame({
    "psi": week_psi.round(3),
    "mae_vs_launch": (weekly["mae"] / launch_mae).round(2),
})
report["verdict"] = [triage(p, r) for p, r in
                     zip(report["psi"], report["mae_vs_launch"])]
print(report.to_string())

fig, ax = plt.subplots(figsize=(7.5, 4.6))
ax.axvline(0.25, color="#6B7785", linestyle=":")
ax.axhline(1.5, color="#6B7785", linestyle=":")
for (x, y, word) in [
        (0.02, 0.04, "healthy"),
        (0.98, 0.04, "early warning"),
        (0.02, 0.96, "error, no input drift"),
        (0.98, 0.96, "drift confirmed")]:
    ax.text(x, y, word, transform=ax.transAxes, fontsize=9,
            color="#6B7785", ha="left" if x < 0.5 else "right",
            va="bottom" if y < 0.5 else "top")
dot_colours = ["#9E3B33" if v.startswith("drift") else "#2C5F79"
               for v in report["verdict"]]
ax.scatter(report["psi"], report["mae_vs_launch"], s=60,
           color=dot_colours, zorder=3)
for week, row in report.iterrows():
    ax.annotate(f"w{week}", (row["psi"], row["mae_vs_launch"]),
                xytext=(6, -3 if week % 2 else 5),
                textcoords="offset points", fontsize=8)
ax.set_xscale("log")
ax.set_xlim(0.02, 2.5)
ax.set_ylim(0.5, 3.8)
ax.set_xlabel("PSI of distance (log scale)")
ax.set_ylabel("MAE / launch MAE")
ax.set_title("The monitoring map: one dot per week", fontsize=10)
ax.spines[["top", "right"]].set_visible(False)
plt.tight_layout()
plt.show()
```

Weeks 1 to 5 fall in **healthy**, including the week 3 false alarm. Weeks 6, 7
and 8 fall in **drift confirmed** and move further up and to the right each
week. The verdict for those three weeks is the same: retrain.

Two states are empty in this scenario. Consider what would place a week in
each.

**Early warning with normal error.** The monsoon arrives and half the orders are
in the rain. The `rain` column shifts, but the model learned what rain does and
remains accurate. The correct action is to watch and not to retrain. L28
measured exactly this case.

**High error with no input drift.** A new traffic rule makes every trip five
minutes slower. The orders look identical and only the answers changed. This is
pure concept drift, and only the error can detect it, which is why monitoring
the inputs alone is insufficient.

The map reads one week at a time. The alert rule from Step 8 adds the
persistence requirement on top of it.

## Step 12: accurate for whom

The audit begins here. The overall MAE for week 8 is a single number. Splitting
it by group establishes whether every group receives the same quality of
service.

Three splits are examined: rain, traffic, and distance.

```python
week8 = live[live["week"] == 8].copy()
week8["error"] = (week8["actual_min"] - week8["predicted_min"]).abs()

print(f"Week 8 overall MAE: {week8['error'].mean():.2f} minutes")
print()

def error_by(frame, column, labels=None):
    table = (frame.groupby(column, observed=True)["error"]
             .agg(["size", "mean"]))
    table.columns = ["orders", "mae"]
    if labels:
        table.index = [labels.get(i, i) for i in table.index]
    return table.round(2)

print("By weather:")
print(error_by(week8, "rain", {0: "dry", 1: "raining"}).to_string())
print()
print("By traffic:")
print(error_by(week8, "traffic_level",
               {1: "light", 2: "medium", 3: "heavy"}).to_string())
print()

week8["trip"] = pd.cut(week8["distance_km"],
                       bins=[0, 5, 12, 100],
                       labels=["short (<5km)", "medium (5-12km)",
                               "long (>12km)"])
print("By distance:")
print(error_by(week8, "trip").to_string())
```

Read the distance table. Short trips are still served well. Long trips, the
ones absent from the training data, are served considerably worse.

The average concealed that entirely. The customers on the wrong side of it are
those furthest from the centre, who are already waiting longest.

## Step 13: plot the audit with the group sizes on the bars

One chart per split. The height is the error and the number on each bar is the
count of orders in that group, because a poor score over four orders is noise
while the same score over seventy is a finding.

A group is drawn red only when it is served at least half again as badly as the
best group in its chart. Smaller differences stay grey, being the ordinary noise
already seen in week 3.

```python
splits = [("rain", {0: "dry", 1: "raining"}),
          ("traffic_level", {1: "light", 2: "medium", 3: "heavy"}),
          ("trip", None)]

fig, axes = plt.subplots(1, 3, figsize=(13, 3.4), sharey=True)
for ax, (column, labels) in zip(axes, splits):
    table = error_by(week8, column, labels)
    worst, best = table["mae"].max(), table["mae"].min()
    unfair = worst > 1.5 * best          # clearly worse, not noise
    colours = ["#9E3B33" if (unfair and v == worst) else "#6B7785"
               for v in table["mae"]]
    bars = ax.bar([str(i) for i in table.index], table["mae"],
                  color=colours)
    ax.bar_label(bars, labels=[f"n={n}" for n in table["orders"]],
                 padding=3, fontsize=8)
    ax.axhline(week8["error"].mean(), color="#2C5F79",
               linestyle="--", linewidth=1)
    ax.set_title(f"by {column}", fontsize=10)
    ax.tick_params(axis="x", labelsize=8)
    ax.spines[["top", "right"]].set_visible(False)
axes[0].set_ylabel("MAE in week 8 (minutes)")
fig.suptitle("Dashed line: the overall average. "
             "It describes nobody in the right-hand chart.",
             fontsize=10)
plt.tight_layout()
plt.show()
```

The left two charts remain grey: rain and traffic are served about equally and
the differences are noise. The right chart is not. One red bar, five times the
height of its neighbours, on the largest group present. That is the chart to put
in front of the person who signs the model off.

## Step 14: establish whether the gap is real or a few bad orders

A bar shows an average, and a small number of extreme cases can raise an
average on their own. Before reporting that a group is served badly, examine
every order in it.

Below, each point is one week 8 order. The box covers the middle half of each
group. The dotted line at plus 10 marks a customer who waited more than ten
minutes longer than promised.

The table then converts MAE into a quantity a manager can act on: the share of
each group's customers who crossed that line.

```python
groups = ["short (<5km)", "medium (5-12km)", "long (>12km)"]
jitter = np.random.default_rng(0)     # spreads the dots; not data

fig, ax = plt.subplots(figsize=(8, 4))
for i, group in enumerate(groups):
    errs = week8.loc[week8["trip"] == group, "signed_error"]
    ax.boxplot(errs, positions=[i], widths=0.5, showfliers=False,
               medianprops={"color": "#6B7785"})
    ax.scatter(i + jitter.uniform(-0.18, 0.18, len(errs)), errs,
               s=12, alpha=0.7, zorder=3,
               color="#9E3B33" if group.startswith("long")
               else "#2C5F79")
ax.axhline(0, color="#6B7785", linewidth=1)
ax.axhline(10, color="#9E3B33", linestyle=":", linewidth=1)
ax.text(0.5, 10.5, "10 min later than promised", fontsize=8,
        color="#9E3B33", ha="center")
ax.set_xticks(range(len(groups)), groups)
ax.set_ylabel("actual - predicted (minutes)")
ax.set_title("Week 8, every order", fontsize=10)
ax.spines[["top", "right"]].set_visible(False)
plt.tight_layout()
plt.show()

late_share = {}
print(f"{'group':<18}{'orders':>7}{'MAE':>7}"
      f"{'avg signed':>12}{'>10 min late':>14}")
for group in groups:
    errs = week8.loc[week8["trip"] == group, "signed_error"]
    late_share[group] = float((errs > 10).mean())
    print(f"{group:<18}{len(errs):>7}{errs.abs().mean():>7.2f}"
          f"{errs.mean():>+12.2f}{late_share[group]:>14.0%}")
```

**Reading the figure.** For short and medium trips the box straddles zero: small
misses, early and late in roughly equal measure. That is a healthy model.

For long trips the whole box sits above zero. One point of 74 lies below it and
more than half lie above the ten minute line. This is not a few unlucky orders
raising an average; the entire group has moved. At 74 orders it is not noise
either.

The final column is the sentence worth stating aloud: short trip customers are
almost never badly late, and long trip customers frequently are. An MAE will not
be remembered outside this room. That sentence will.

## Step 15: the audit checklist

The final check before the repair. Before any model goes live, and every quarter
afterwards, somebody should be able to answer each of these questions.

```python
checklist = [
    ("Who is worst served?",
     "split the error by every group you can name"),
    ("How would we know it broke?",
     "name the metric, the threshold and who gets the alert"),
    ("How fast would we know?",
     "inputs are instant; errors wait for the truth to arrive"),
    ("What was it trained on?",
     "dates, row count, and which groups were thin or missing"),
    ("Can we explain one prediction?",
     "to a customer, not to a data scientist"),
    ("Can we turn it off?",
     "what serves traffic while the model is switched out"),
    ("What did we log?",
     "enough to investigate, nothing that identifies a person"),
    ("Who signed it off?",
     "a name and a date, not a team"),
]

for i, (question, how) in enumerate(checklist, 1):
    print(f"{i}. {question}")
    print(f"   -> {how}")

print()
print("Our model fails number 1 today: long trips are served")
print("far worse than short ones. The honest fix is to retrain")
print("on data that includes the new suburb -- not to report")
print("the average and hope. That retrain is Practical P14.")
```

## Step 16: write it down, the incident note

Charts convince the people present. A short written note convinces those who
were not: the manager, the next team, and the author in three months.

A usable note answers five questions in plain words: what was observed, when,
who is affected, why it happened, and what will be done. Every number below is
read from a variable computed earlier in this notebook, so the note cannot
disagree with the charts.

This is also the shape of the written answers required in the submission.

```python
trip_table = error_by(week8, "trip")
long_row, short_row = (trip_table.loc["long (>12km)"],
                       trip_table.loc["short (<5km)"])

lines = [
    f"INCIDENT NOTE -- delivery-time model {metrics['model_version']}",
    "",
    f"WHAT    Error rose from {launch_mae:.2f} min at launch "
    f"to {weekly['mae'].iloc[-1]:.2f} min in week 8.",
    f"WHEN    Inputs shifted in week {first_shift} "
    f"(PSI {week_psi[first_shift]:.2f}); "
    f"the error alert fired in week {first_alert}.",
    f"WHO     Long trips (>12 km): MAE {long_row['mae']:.2f} min "
    f"on {int(long_row['orders'])} orders, "
    f"against {short_row['mae']:.2f} for short trips.",
    f"        {late_share['long (>12km)']:.0%} of long-trip "
    f"customers waited 10+ min longer than promised.",
    f"WHY     No training trip was longer than "
    f"{longest_trained:.1f} km. Beyond that, the model",
    f"        under-predicts by {beyond_bias:.1f} min on average; "
    f"inside it, MAE is still {inside_mae:.2f}.",
    "ACTION  Retrain on data that includes the new suburb (P14).",
    "        Until then, add a safety margin to long-trip estimates.",
]
note = "\n".join(lines)

print(note)
(WORK / "incident_note.txt").write_text(note, encoding="utf-8")
print("saved to", WORK / "incident_note.txt")
```

Read the note as somebody who has never seen this notebook. It names a group, a
size, a cause and an action, and it fits on a phone screen. Note what it does
not say: that the model's MAE is 6.50. That number, on its own, describes
nobody.
